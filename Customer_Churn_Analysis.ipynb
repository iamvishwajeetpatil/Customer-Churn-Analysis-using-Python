# Replacing the empty values with 0 and changeing the datatype from object to float

```python
df["TotalCharges"] = df["TotalCharges"].replace(" ","0")
df["TotalCharges"] = df["TotalCharges"].astype("float")
```

# Converted 0 and 1 value of SeniorCitizen cloumn to No and yes value ease to understand

```python
def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"
df["SeniorCitizen"] = df["SeniorCitizen"].apply(conv)
```

# show the Churn on the graph

```python
plt.figure(figsize = (4,4))
ax = sns.countplot(data=df, x='Churn')
ax.bar_label(ax.containers[0])
plt.show()
```

```python
plt.figure(figsize = (3,3))
gb = df.groupby("Churn").agg({'Churn':"count"})
plt.pie(gb['Churn'], labels = gb.index, autopct = "%1.2f%%")
plt.show()
```

# From the given pie chart we can conclude that 26.54% of our customer have chured out 

# Now lets find out the reasone fro it 

```python
plt.figure(figsize = (4,4))
ax = sns.countplot( x= "gender", data = df, hue = "Churn")
ax.bar_label(ax.containers[0])
plt.title("Churn by Gender")
plt.show()
```

```python

# Create crosstab
ct = pd.crosstab(df["SeniorCitizen"], df["Churn"])

# Convert to percentage
ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100

# Plot stacked bar chart
ax = ct_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(5,5)
)

# Add percentage labels
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.1f%%',
        label_type='center',
        fontsize=10,
        color='white'
    )

plt.title("Churn by SeniorCitizen (%)")
plt.xlabel("SeniorCitizen")
plt.ylabel("Percentage")
plt.legend(title="Churn")
plt.xticks(rotation=0)

plt.show()
```

```python

# List of columns
cols = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# Create subplot layout
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(18, 15))

# Flatten axes for easy looping
axes = axes.flatten()

# Loop through columns and create countplots
for i, col in enumerate(cols):
    sns.countplot(
        data=df,
        x=col,
        hue="Churn",
        ax=axes[i]
    )

    axes[i].set_title(f"{col} vs Churn")
    axes[i].tick_params(axis='x', rotation=20)

# Adjust layout
plt.tight_layout()

# Show plots
plt.show()
```

Summary of Service Features vs Churn

From the count plots, several clear churn patterns emerge:

Key Insights
InternetService
Customers using Fiber optic internet show the highest churn rate.
DSL users churn significantly less.
Customers with No internet service have very low churn.

Insight: Fiber optic customers may be facing issues related to pricing, service quality, or customer expectations.

OnlineSecurity
Customers without OnlineSecurity have a very high churn count.
Customers who use OnlineSecurity are much more likely to stay.

Insight: Security-related add-on services improve customer retention.

OnlineBackup
Churn is much higher among customers who do not use OnlineBackup.
Customers with backup services churn less.

Insight: Additional utility services increase customer stickiness.

DeviceProtection
Users without DeviceProtection churn significantly more.
Customers subscribed to protection plans tend to remain loyal.

Insight: Protection and warranty-like services positively impact retention.

TechSupport
One of the strongest indicators of churn.
Customers without TechSupport have extremely high churn.
Customers with TechSupport churn far less.

Insight: Good customer support is a major retention factor.

StreamingTV & StreamingMovies
Churn difference is less dramatic compared to support/security services.
Both Yes/No groups show moderate churn levels.

Insight: Entertainment services alone do not strongly influence retention.

PhoneService & MultipleLines
These features show relatively smaller churn differences.
Having phone services or multiple lines does not strongly impact churn behavior.

Insight: Core communication services are not major churn drivers.

Overall Business Conclusion

The strongest churn drivers are:

Fiber optic internet users
Lack of TechSupport
Lack of OnlineSecurity
Lack of DeviceProtection
Lack of OnlineBackup

Customers using additional support and protection services are significantly more likely to stay with the company.
